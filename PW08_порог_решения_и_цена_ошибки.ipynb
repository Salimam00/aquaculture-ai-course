```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import precision_recall_fscore_support
df=pd.read_csv("../../sturgeon_risk_classification.csv");X=df.drop(columns="risk");y=df.risk
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.25,stratify=y,random_state=42)
m=LogisticRegression(max_iter=2000,class_weight="balanced").fit(Xtr,ytr); pr=m.predict_proba(Xte)[:,1]
for tau in [.3,.5,.7]:
 p,r,f,_=precision_recall_fscore_support(yte,pr>=tau,average="binary",zero_division=0);print(tau,p,r,f)
```
