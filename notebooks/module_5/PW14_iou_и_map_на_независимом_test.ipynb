## Оценивание

Используйте общую аналитическую рубрику из `assessments/rubric.md`.


```python
import pandas as pd, numpy as np
a=pd.read_csv("../../data/cv/annotations.csv"); test=a[a.split=="test"].copy()
def iou(a,b):
 x1=max(a[0],b[0]);y1=max(a[1],b[1]);x2=min(a[2],b[2]);y2=min(a[3],b[3]); inter=max(0,x2-x1)*max(0,y2-y1);u=(a[2]-a[0])*(a[3]-a[1])+(b[2]-b[0])*(b[3]-b[1])-inter;return inter/u
r=np.random.default_rng(42); scores=[]
for _,q in test.iterrows():
 gt=q[["x1","y1","x2","y2"]].to_numpy(float);pred=gt+r.normal(0,3,4);scores.append(iou(gt,pred))
print("Mean IoU",np.mean(scores),"Recall@IoU.50",np.mean(np.array(scores)>=.5))
print("Примечание: это учебная симуляция, не результат YOLO и не mAP.")
```
