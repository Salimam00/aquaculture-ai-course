```python
import pandas as pd
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures,StandardScaler
from sklearn.linear_model import Ridge
from sklearn.model_selection import KFold,cross_val_score
df=pd.read_csv("../../data/fish_growth_dataset.csv");X=df[["age_weeks","mean_water_temp_C","daily_feed_ration_g"]];y=df.weight_g
cv=KFold(5,shuffle=True,random_state=42)
for d in [1,2,3]:
 m=make_pipeline(PolynomialFeatures(d),StandardScaler(),Ridge(alpha=1)); s=-cross_val_score(m,X,y,cv=cv,scoring="neg_mean_absolute_error")
 print(d,s.mean(),s.std())
```
